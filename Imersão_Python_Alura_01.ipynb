```python
df.columns

# Mapeamento dos nomes das colunas para o português do Brasil
colunas_traduzidas = {
    'ano': 'ano',
    'nivel_experiencia': 'nivel_experiencia',
    'tipo_emprego': 'tipo_emprego',
    'cargo': 'cargo',
    'salario': 'salario',
    'moeda_salario': 'moeda',
    'salario_em_usd': 'usd',
    'residencia_funcionario': 'residencia',
    'percentual_remoto': 'percentual_remoto',
    'localizacao_empresa': 'localizacao_empresa',
    'porte_empresa': 'porte_empresa'
}

# Renomear as colunas do DataFrame
df = df.rename(columns=colunas_traduzidas)

# Exibir as novas colunas para verificar a tradução
print('Novas colunas do DataFrame:')
print(df.columns)
```

```python
df['nivel_experiencia'].value_counts()
```

```python
df['tipo_emprego'].value_counts()
```

```python
df['percentual_remoto'].value_counts()
```

```python
df['porte_empresa'].value_counts()
```

```python
mapeamento_nivel_experiencia = {
    'SE': 'Senior',
    'MI': 'Pleno',
    'EN': 'Junior',
    'EX': 'Executivo'
}
df['nivel_experiencia'] = df['nivel_experiencia'].map(mapeamento_nivel_experiencia)
print(df['nivel_experiencia'].value_counts())
```

```python
mapeamento_tipo_emprego = {
    'FT': 'Tempo Integral',
    'CT': 'Contrato',
    'PT': 'Meio Período',
    'FL': 'Freelancer'
}
df['tipo_emprego'] = df['tipo_emprego'].map(mapeamento_tipo_emprego)
print(df['tipo_emprego'].value_counts())
```

```python
mapeamento_porte_empresa = {
    'M': 'Médio',
    'L': 'Grande',
    'S': 'Pequeno'
}
df['porte_empresa'] = df['porte_empresa'].map(mapeamento_porte_empresa)
print(df['porte_empresa'].value_counts())
```

```python
mapeamento_percentual_remoto = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Totalmente Remoto'
}
df['percentual_remoto'] = df['percentual_remoto'].map(mapeamento_percentual_remoto)
print(df['percentual_remoto'].value_counts())
```
